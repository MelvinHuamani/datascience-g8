```sql
USE tienda
```

```sql
--Ejercicio 1

SELECT
    nombre,
    precio
FROM productos
ORDER BY precio DESC
```

```sql
--Ejercicio 2

SELECT *
FROM clientes
WHERE pais IN ('ES', 'MX')
  AND nombre LIKE 'M%'
```

```sql
--Ejercicio 3

SELECT *
FROM productos
WHERE precio BETWEEN 20 AND 60
  AND stock > 0;
```

```sql
--Ejercicio 4

SELECT
    nombre,
    COALESCE(telefono, 'sin dato') AS telefono
FROM clientes;
```

```sql
--Ejercicio 5

SELECT
    nombre,
    precio,
    precio * 1.21 AS precio_IVA
FROM productos
ORDER BY precio DESC
LIMIT 10;
```

```sql
--Ejercicio 6

SELECT
    nombre,
    precio,
    CASE
        WHEN precio > 300 THEN 'caro'
        WHEN precio > 50 THEN 'medio'
        ELSE 'barato'
    END AS clasificacion
FROM productos;
```

```sql
--Ejercicio 7

SELECT
    DISTINCT pais AS cantidad_paises
FROM clientes;
```
