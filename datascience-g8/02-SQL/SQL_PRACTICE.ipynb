```sql
CREATE OR REPLACE TABLE categorias (
    id      INT   NOT NULL,
    nombre STRING NOT NULL
);

CREATE OR REPLACE TABLE productos (
    id      INT   NOT NULL,
    nombre STRING NOT NULL,
    precio NUMERIC NOT NULL,
    stock INT,
    categoria_id INT NOT NULL
);

CREATE OR REPLACE TABLE detalle_pedidos (
    pedido_id      INT   NOT NULL,
    producto_id INT NOT NULL,
    cantidad INT,
    precio_unit NUMERIC
);

CREATE OR REPLACE TABLE pedidos (
    id      INT   NOT NULL,
    cliente_id INT NOT NULL,
    fecha DATE,
    total NUMERIC
);

CREATE OR REPLACE TABLE clientes (
    id      INT   NOT NULL,
    nombre STRING NOT NULL,
    email STRING,
    pais STRING,
    telefono STRING
);

```

```sql
SELECT
    LENGTH(nombre) AS longitud_nombre,
    UPPER(nombre) AS nombre_mayusculas,
    LOWER(nombre) AS nombre_minusculas,
    CONCAT(nombre, ' ; ', email) AS nombre_email,
    CONCAT(nombre, email) AS nombre_email_2,
    'UCSP' AS universidad,
    LENGTH(' UCSP ') AS longitud_universidad,
    TRIM(' UCSP ') AS universidad_trim,
    LENGTH(TRIM(' UCSP ')) AS longitud_universidad_trim,
    pais,
    REPLACE(pais, 'ES', 'España') AS pais_reemplazado,
    REPLACE(nombre, 'o', 'x') AS nombre_reemplazado,
    nombre,
    SUBSTRING(nombre, 1, 3) AS nombre_tres_primeras_letras
FROM tienda.clientes;
```

```sql
SELECT
    'oreo' AS galleta,
    4.595 AS precio,
    ROUND(4.595, 1) AS precio_redondeado;
    
```

```sql
SELECT
    pais,
    CASE
        WHEN pais = 'ES' THEN 'Espana'
        WHEN pais = 'MX' THEN 'Mexico'
        WHEN pais = 'PE' THEN 'Peru'
        WHEN pais = 'CO' THEN 'Colombia'
        WHEN pais = 'CL' THEN 'Chile'
        ELSE 'Argentina'
    END AS nombre_pais
FROM clientes;
```

```sql
SELECT
    nombre,
    precio,
    CASE
        WHEN precio > 300 THEN 'caro'
        WHEN precio > 50 THEN 'medio'
        ELSE 'barato'
    END AS clasificacion
FROM productos;
```

```sql
SELECT *,
    COALESCE(categoria_id, 0) AS categoria_id_imputada
    --NVL(categoria_id, 0) AS categoria_id_imputada
FROM productos;
```
