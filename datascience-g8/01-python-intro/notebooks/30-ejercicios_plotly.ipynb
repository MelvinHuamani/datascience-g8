```python
# 1. Gráfico de dispersión básico
# Construye un gráfico de dispersión que muestre la relación entre `sepal_length` y `sepal_width`. Diferencia
# cada observación según la especie mediante el parámetro `color`. Agrega un título descriptivo y etiquetas
# claras para ambos ejes.
fig = px.scatter(
    data_frame=df,
    x="sepal_length",
    y="sepal_width",
    color="species",
    title="Longitud vs ancho del sépalo por especie",
    labels={
        "sepal_length": "Longitud del sépalo (cm)",
        "sepal_width": "Ancho del sépalo (cm)",
        "species": "Especie"
    }
)
fig.show()
```

```python
# 2. Histograma comparativo por especie
# Representa la distribución de `petal_length` mediante un histograma. Usa 20 bins, diferencia las especies
# con `color`, configura `barmode="overlay"` y aplica una opacidad que permita comparar las distribuciones
# superpuestas.

fig = px.histogram(
    data_frame=df,
    x="petal_length",
    color="species",
    nbins=20,
    barmode="overlay",
    opacity=0.65,
    title="Distribución del largo del pétalo por especie",
    labels={
        "petal_length": "Longitud del pétalo (cm)",
        "species": "Especie"
    }
)

fig.show()
```

```python
# 4. Violin plot del largo del sépalo
# Construye un gráfico de violín para comparar `sepal_length` entre especies. Incluye dentro del violín un
# boxplot y muestra todos los puntos. Explica qué información adicional aporta el violín respecto al boxplot.

fig = px.violin(
    data_frame=df,
    x="species",
    y="sepal_length",
    color="species",
    points="all",
    box=True,
    title="Distribucion del largo del sépalo por especie",
    labels={
        "species": "Especie",
        "sepal_length": "Longitud del sépalo (cm)"
    }
)

fig.show()
```

```python
# 5. Promedio de largo del sépalo por especie
# Calcula el promedio de `sepal_length` por especie con Pandas y visualízalo mediante un gráfico de barras.

promedio_sepalo = (
    df.groupby("species", as_index=False)["sepal_length"].mean()
)
# promedio_sepalo

fig = px.bar(
    promedio_sepalo,
    x="species",
    y="sepal_length",
    text_auto=".3f",
    title="Promedio de longitud del sépalo por especie",
    labels={"species": "Especie", "sepal_length": "Promedio (cm)"}
)

fig.show()
```

```python
promedio_sepalo
```

```python
# 6. Comparación de promedios de las cuatro medidas
# Calcula, para cada especie, el promedio de `sepal_length`, `sepal_width`, `petal_length` y `petal_width`.
# Crea un gráfico de barras agrupadas que permita comparar las
# cuatro variables entre especies.

promedios = df.groupby("species", as_index=False)[["sepal_length", "sepal_width", "petal_length", "petal_width"]].mean()
promedios

fig = px.bar(
    promedios,
    x="species",
    y=["sepal_length", "sepal_width", "petal_length", "petal_width"],
    barmode="group",
    title="Promedios de las cuatro medidas por especie",
    labels={"species": "Especie", "value": "Promedio (cm)"}
)

fig.show()
```

```python
# 7. Scatter multivariable
# Grafica `petal_length` frente a `petal_width`. Usa la especie como `color`, el largo del sépalo como tamaño
# de los puntos (`size`) y la especie también como símbolo (`symbol`). Agrega información adicional al tooltip
# mediante `hover_data`.

fig = px.scatter(
    data_frame=df,
    x="petal_length",
    y="petal_width",
    color="species",
    size="sepal_length",
    symbol="species",
    hover_data=["sepal_width", "species"],
    title="Relación entre medidas del pétalo",
    labels={
        "petal_length": "Longitud del pétalo (cm)",
        "petal_width": "Ancho del pétalo (cm)",
        "sepal_length": "Longitud del sépalo (cm)",
        "species": "Especie",
        "sepal_width": "Ancho del sépalo (cm) xxxxxx"
    }
)

fig.show()
```

```python
# 8. Comparación mediante facetas
# Crea un gráfico de dispersión de `sepal_length` frente a `petal_length` y separa cada especie en una faceta
# distinta usando `facet_col`. Mantén una escala comparable para poder observar diferencias entre especies.
fig = px.scatter(
    data_frame=df,
    x="sepal_length",
    y="petal_length",
    facet_col="species",
    color="species",
    title="Longitud del sépalo vs pétalo",
    labels={
        "sepal_length": "Longitud del sépalo (cm)",
        "petal_length": "Longitud del pétalo (cm)",
        "species": "Especie"
    }
)

fig.show()
```

```python
# 9. Gráfico de dispersión 3D
# Construye un gráfico 3D usando `sepal_length` en X, `sepal_width` en Y y `petal_length` en Z. Colorea los
# puntos por especie y agrega `petal_width` al tooltip. Explora si las especies forman grupos diferenciables.

fig = px.scatter_3d(
    data_frame=df,
    x="sepal_length",
    y="sepal_width",
    z="petal_length",
    color="species",
    hover_data=["petal_width"],
    title="Grafico scatter 3D"
)
fig.show()
```
