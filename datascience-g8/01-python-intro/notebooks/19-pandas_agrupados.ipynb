```python

import pandas as pd

ventas = pd.read_csv(
    r"C:\Users\melvi\Documents\datascience-g8\01-python-intro\data\ventas.csv"
)

ventas["importe"] = ventas["unidades"] * ventas["precio_unitario"]

ventas
```

```python
ventas.groupby("region")["importe"].sum()
```

```python
ventas[ventas["region"] == "Sur"]["importe"].sum()
```

```python
ventas[ventas["region"] == "Sur"]["importe"].mean()
```

```python
ventas[ventas["region"] == "Sur"].shape
```

```python
ventas.groupby("region")["importe"].agg(
    importe_total="sum",
    importe_promedio="mean",
    total_registros="count"
)
```

```python
ventas.groupby(["region", "categoria"])["importe"].agg(
    importe_total="sum",
    importe_promedio="mean",
    total_registros="count"
)
```

```python
ventas.groupby("region").agg({
    "importe": ["sum", "mean", "max"],
    "unidades": ["sum"],
    "producto": ["nunique"]
})
```
