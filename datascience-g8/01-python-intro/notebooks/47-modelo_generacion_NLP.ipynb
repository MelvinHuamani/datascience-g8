```python
from transformers import pipeline
import torch

if torch.backends.mps.is_available():
    DEVICE = "mps"
elif torch.cuda.is_available():
    DEVICE = "cuda"
else:
    DEVICE = "cpu"

print(f"DEVICE: {DEVICE}")

clasificador = pipeline("sentiment-analysis",
                        dtype=torch.float16 if DEVICE in ("cuda", "mps") else torch.float32,
                        device=DEVICE,
                        model="nlptown/bert-base-multilingual-uncased-sentiment")
```

```python
print(clasificador("Me encantó la película"))
```

```python
print(clasificador("La carta de esta restaurante es muy mala, no tiene variedad"))
```

```python
print(clasificador("Genial, se rompió al segundo día"))
```

```python
print(clasificador("No está nada mal para el precio"))
```
