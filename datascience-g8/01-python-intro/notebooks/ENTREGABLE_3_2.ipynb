```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\melvi\Documents\datascience-g8\01-python-intro\data\iris_dataset.csv")
```

```python
#9. Histograma: longitud del sépalo
plt.figure(figsize=(8, 4))

plt.hist(df["sepal_length_cm"], bins=10, edgecolor="black")

plt.title("Histograma de longitud ")
plt.xlabel("Longitud (cm)")
plt.ylabel("Frecuencia")

plt.show()
```

```python
#10. Histograma y bins

fig, axes = plt.subplots(1, 2, figsize=(8, 4))

#5 bins
axes[0].hist(df["petal_length_cm"], bins=5, edgecolor="black")
axes[0].set_title("Longitud 5 bins")
axes[0].set_xlabel("Longitud (cm)")
axes[0].set_ylabel("Frecuencia")

#20 bins
axes[1].hist(df["petal_length_cm"], bins=20, edgecolor="black")
axes[1].set_title("Longitud 20 bins")
axes[1].set_xlabel("Longitud (cm)")
axes[1].set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

```python
#11. Boxplot por especie
plt.figure(figsize=(8, 4))

df.boxplot(
    column="sepal_length_cm",
    by="species",
    grid=False,
    patch_artist=True
)

plt.title("Longitud por especie")
plt.suptitle("")
plt.xlabel("Especie")
plt.ylabel("Longitud (cm)")

plt.show()
```

```python
#12. Boxplot de variables numéricas
plt.figure(figsize=(8, 4))

df.boxplot(
    column=[
        "sepal_length_cm",
        "sepal_width_cm",
        "petal_length_cm",
        "petal_width_cm"
    ],
    grid=False,
    patch_artist=True
)

plt.title("Boxplot variables/numericas")
plt.xlabel("Variables")
plt.ylabel("Valores")

plt.xticks(
    [1, 2, 3, 4],
    ["Sepal L.", "Sepal W.", "Petal L.", "Petal W."]
)

plt.show()
```

```python
#13. Subplots 1x2
fig, axes = plt.subplots(1, 2, figsize=(8, 4))

#Scatter
colores = ["red", "green", "blue"]

for especie, color in zip(["setosa", "versicolor", "virginica"], colores):
    datos = df[df["species"] == especie]
    axes[0].scatter(
        datos["sepal_length_cm"],
        datos["sepal_width_cm"],
    )

axes[0].set_title("Longitud/ancho")
axes[0].set_xlabel("Longitud (cm)")
axes[0].set_ylabel("Ancho (cm)")

#Histograma
axes[1].hist(
    df["sepal_length_cm"],
    bins=10,
    edgecolor="black",
    color="blue"
)

axes[1].set_title("Distribución de la longitud")
axes[1].set_xlabel("Longitud (cm)")
axes[1].set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```
