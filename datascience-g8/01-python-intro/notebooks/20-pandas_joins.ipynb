```python
## CONCATENADO

import pandas as pd

enero = pd.DataFrame({
    "mes": ["Enero"],
    "num_mes": [1]
})

febrero = pd.DataFrame({
    "mes": ["Febrero"],
    "num_mes": [2]
})

marzo = pd.DataFrame({
    "mes": ["Marzo"],
    "num_mes": [3]
})
```

```python
pd.concat([enero, febrero, marzo], axis=1)
```

```python

pd.concat([enero, febrero, marzo], axis=0)
```
