CONVERSION LISTA A NUMPY ARRAY

```python
lista = [1, 2, 3, 5]

arr_lista = np.array(lista)
print(arr_lista)
print(type(arr_lista))

arr_lista * 2
```

```python
arr_lista.dtype
```

```python
lista_decimales = [1.0, 2.1, 2.4, 5.6]
arr_lista = np.array(lista_decimales)
arr_lista.dtype
```

```python
arr_lista = np.array(lista, dtype='float64')
print(arr_lista)
print(arr_lista.dtype)
```

CONVERTIR TIPO DE DATOS DE MI ARRAY

```python
lista = [1, 2, 3, 5]

arr_lista = np.array(lista)
print(arr_lista)
arr_lista.dtype
```

```python
arr_float = arr_lista.astype('float64')
print(arr_float)
arr_float.dtype
```

```python
arr_str = arr_lista.astype('str')
print(arr_str)
arr_str.dtype
```

DIMENSIONES

```python
# SCALAR
scalar = np.array(50)
print(scalar)
print(type(scalar))
print(scalar.ndim)
```

```python
# VECTOR
vector = np.array([1, 2, 3, 4])
print(vector)
print(vector.ndim)
```

```python
# MATRIZ
matrix = np.array([[1,2,3], [4,5,6]])
print(matrix)
print(matrix.ndim)
```
