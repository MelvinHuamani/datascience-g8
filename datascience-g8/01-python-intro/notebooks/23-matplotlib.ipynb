```python
import matplotlib

print(matplotlib.__version__)
```

```python
import numpy as np

x = np.linspace(0, 5, 10)
y = x ** 2
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y, 'b')
plt.show()
```

```python
plt.plot(x, y, 'ro')
```

```python
plt.plot(x, y, 'ko-')
```

```python
plt.plot(x, y, 'gD:')
```

```python
plt.plot(x, y, color="#F59F27", linestyle='--', marker='o')
```

```python
plt.plot(x, y, color="#F59F27", linestyle='--', linewidth=2, marker='o', markeredgecolor='b', markerfacecolor='b')
```
