```python
# EJERCICIO 1

# Actividades
# 1. Convierte la lista en un array de NumPy y reorganízalo como una matriz de 4 × 6.
# 2. Muestra su tipo de dato, número de dimensiones y forma.
# 3. Calcula el total vendido en cada turno y el total vendido de cada producto.
# 4. Identifica el turno con mayor cantidad total de ventas.
# 5. Obtén una copia con todas las ventas incrementadas en 2 unidades y verifica que la matriz original no
# cambió.

ventas = [
    18, 12, 25, 10, 8, 15,
    20, 14, 22, 13, 11, 17,
    16, 19, 27, 9, 12, 18,
    24, 17, 21, 15, 10, 20
]

ventas_matriz = np.array(ventas)
# 1
ventas_matriz = ventas_matriz.reshape((4, 6))

# 2
print(ventas_matriz.dtype) # tipo de dato
print(ventas_matriz.ndim) # numero de dimensiones
print(ventas_matriz.shape) # filas x columnas
print('-'*100)

# 3
venta_total_turno = ventas_matriz.sum(axis=1)
print(venta_total_turno)
print('-'*100)

venta_total_producto = ventas_matriz.sum(axis=0)
print(venta_total_producto)
print('-'*100)

# 4
print(venta_total_turno.argmax())
print('-'*100)

# 5
ventas_incrementadas = ventas_matriz.copy()
ventas_incrementadas += 2
print(ventas_matriz)
print(ventas_incrementadas)
```

```python
# Ejercicio 2

stock = np.array(
    [
        [25, 18, 30, 12, 20],
        [14, 22, 16, 28, 19],
        [35, 17, 24, 15, 21]
    ]
)

ventas = np.array(
    [
        [10, 20, 12, 8, 7],
        [16, 10, 18, 9, 20],
        [15, 19, 8, 17, 11]
    ]
)

# Actividades
# 1. Calcula el inventario restante mediante una operación entre matrices.
stock_restante = stock - ventas
print(stock_restante)
print('-'*100)

# 2. Localiza los valores negativos, que representan ventas mayores al stock disponible.
mascara = stock_restante < 0
print(stock_restante[mascara])
print('-'*100)

# 3. Crea una copia corregida en la que los valores negativos sean reemplazados por 0.
stock_restante_corregido = stock_restante.copy()
stock_restante_corregido[mascara] = 0
print(stock_restante_corregido)
print('-'*100)

# 4. Calcula el stock restante total por categoría antes y después de la corrección.
stock_restante_antes = stock_restante.sum(axis=1)
print(stock_restante_antes)

stock_restante_despues = stock_restante_corregido.sum(axis=1)
print(stock_restante_despues)
print('-'*100)

# 5. Determina la categoría con mayor stock disponible después de la corrección.
print(stock_restante_despues.argmax() + 1)
```

```python
# Ejercicio 3

precios = np.array([
    [4.5, 6.0, 3.5, 8.0],
    [4.7, 5.8, 3.6, 8.2],
    [4.6, 6.1, 3.4, 7.9]
])

unidades = np.array([
    [30, 18, 42, 15],
    [25, 20, 35, 17],
    [28, 22, 40, 14]
])

# 1. Calcula la facturación por producto y sucursal mediante multiplicación elemento a elemento.
total = precios * unidades
print(total)

# 2. Calcula la facturación total de cada sucursal y la facturación total de cada producto.
total_sucursal = total.sum(axis=1)
total_producto = total.sum(axis=0)
print(total_sucursal)
print(total_producto)
print('-'*100)

# 3. Identifica la sucursal con mayor facturación.
print(total_sucursal.argmax() + 1)
print('-'*100)

# 4. Encuentra el valor máximo de facturación individual y su posición dentro de la matriz.
valor_maximo = total.max()
print(valor_maximo)

for fila in range(total.shape[0]):
    for columna in range(total.shape[1]):
        if valor_maximo == total[fila, columna]:
            print(fila, columna)
print('-'*100)

# 5. Obtén la transpuesta de la matriz de facturación e interpreta qué representan ahora sus filas.
print(total.T)
```
