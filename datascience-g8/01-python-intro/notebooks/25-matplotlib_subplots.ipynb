```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
x = np.linspace(0, 5, 11)
y1 = x ** 2
y2 = x ** 3

plt.subplot(1, 2, 1)
plt.plot(x, y1, 'bo-')
plt.title('Función Cuadrática')

plt.subplot(1, 2, 2)
plt.plot(x, y2, 'go-')
plt.title('Función Cúbica')

plt.tight_layout(pad=0.01, h_pad=0.5, w_pad=10)
plt.show()
```

```python
x = np.linspace(0, 5, 11)
y1 = x ** 2
y2 = x ** 3

frutas = ['Manzanas', 'Plátanos', 'Peras']
porcentajes = [30, 50, 20]


categorias = ['A', 'B', 'C', 'D']
valores = [40, 70, 30, 90]

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(x, y1, 'bo-')
plt.title('Función Cuadrática')

plt.subplot(2, 2, 2)
plt.plot(x, y2, 'go-')
plt.title('Función Cúbica')

plt.subplot(2, 2, 3)
plt.pie(porcentajes, labels=frutas, autopct='%1.1f%%')
plt.title('Gráfico de pastel')

plt.subplot(2, 2, 4)
plt.bar(categorias, valores, width=0.5, label=categorias, color=['red', 'black', 'blue', 'green'])
plt.xlabel('Categoria')
plt.ylabel('Frecuencia')
plt.title('Gráfico de barras verticales')

plt.tight_layout(w_pad=5, h_pad=5)
plt.show()
```
