```python
def extract_contenido(ruta: str) -> list[str]:

    lista = []

    with open(ruta, "r") as archivo:

        lineas = archivo.readlines()

        for linea in lineas:
            linea = linea.strip()
            if "Instrucciones sugeridas para el ejercicio" in linea:
                break
            if linea == "":
                continue
            if "Registro" in linea:
                continue
            if "ID" in linea:
                continue
            lista.append(linea)

    return lista


contenido = extract_contenido(
    r"C:\Users\melvi\Documents\datascience-g8\01-python-intro\data\ventas_polos_aqp.txt"
)
```

```python
class Venta:

    def __init__(
        self,
        ID: str,
        producto: str,
        talla: str,
        color: str,
        cliente: str,
        ciudad: str,
        cantidad: int,
        precio_unitario: float,
        metodo_pago: str,
        estado: str
    ) -> None:
        self.ID = ID
        self.producto = producto
        self.talla = talla
        self.color = color
        self.cliente = cliente
        self.ciudad = ciudad
        self.cantidad = cantidad
        self.precio_unitario = precio_unitario
        self.metodo_pago = metodo_pago
        self.estado = estado

    def total(self) -> float:
        return self.cantidad * self.precio_unitario

class Registro_Ventas:

    def __init__(self):
        self.ventas = []

    def agregar_venta(self, venta: Venta):
        self.ventas.append(venta)

    def contar_estados(self):

        entregadas = 0
        pendientes = 0
        canceladas = 0

        for venta in self.ventas:
            if venta.estado.lower() == "entregado":
                entregadas += 1
            elif venta.estado.lower() == "pendiente":
                pendientes += 1
            elif venta.estado.lower() == "cancelado":
                canceladas += 1

        return entregadas, pendientes, canceladas

    def ingreso_total(self):

        total = 0

        for venta in self.ventas:
            if venta.estado.lower() == "entregado":
                total += venta.total()

        return total


    def producto_mas_vendido(self):

        productos = {}

        for venta in self.ventas:
            if venta.producto not in productos:
                productos[venta.producto] = 0
            productos[venta.producto] += venta.cantidad

        mayor_producto = ""
        mayor_cantidad = 0

        for producto, cantidad in productos.items():
            if cantidad > mayor_cantidad:
                mayor_producto = producto
                mayor_cantidad = cantidad

        return mayor_producto, mayor_cantidad


    def ventas_arequipa(self):

        contador = 0

        for venta in self.ventas:
            if venta.ciudad.lower() == "arequipa":
                contador += 1

        return contador


    def metodo_pago_mas_usado(self):

        metodos = {}

        for venta in self.ventas:
            if venta.metodo_pago not in metodos:
                metodos[venta.metodo_pago] = 0
            metodos[venta.metodo_pago] += 1

        metodo = ""
        mayor = 0

        for m, cantidad in metodos.items():
            if cantidad > mayor:
                mayor = cantidad
                metodo = m

        return metodo, mayor


    def resumen_productos(self):

        resumen = {}

        for venta in self.ventas:
            if venta.producto not in resumen:
                resumen[venta.producto] = {
                    "unidades": 0,
                    "ingreso": 0
                }

            resumen[venta.producto]["unidades"] += venta.cantidad
            resumen[venta.producto]["ingreso"] += venta.total()

        return resumen


# Crear Ventas

registro = Registro_Ventas()

for linea in contenido:
    datos = linea.split(",")

    if len(datos) != 10:
        continue

    venta = Venta(
        datos[0].strip(),
        datos[1].strip(),
        datos[2].strip(),
        datos[3].strip(),
        datos[4].strip(),
        datos[5].strip(),
        int(datos[6].strip()),
        float(datos[7].strip()),
        datos[8].strip(),
        datos[9].strip()
    )

    registro.agregar_venta(venta)


#Total vendido

print("Total vendido por cada registro\n")

for venta in registro.ventas:
    print(
        venta.ID,
        ":",
        venta.producto,
        venta.total()
    )


#Estados de venta

entregadas, pendientes, canceladas = registro.contar_estados()

print("\nCantidad de ventas por estado\n")

print("Entregadas:", entregadas)

print("Pendientes:", pendientes)

print("Canceladas:", canceladas)


#Ingreso por total entregado

print("\nIngreso total de ventas entregadas\n")

print(registro.ingreso_total())


#Producto mas vendido

producto, cantidad = registro.producto_mas_vendido()

print("\nProducto mas vendido\n")

print(producto, cantidad)


#Ventas en Arequipa

print("\nVentas realizadas en Arequipa\n")

print(registro.ventas_arequipa())



#Metodo de pago mas usado

metodo, veces = registro.metodo_pago_mas_usado()

print("\nMetodo de pago mas usado\n")

print(metodo, veces)



#Resumen por producto

print("\nResumen por producto\n")

resumen = registro.resumen_productos()

for producto, datos in resumen.items():

    print(
        producto,
        "/Unidades:",
        datos["unidades"],
        "/Ingreso:",
        datos["ingreso"]
    )
```
